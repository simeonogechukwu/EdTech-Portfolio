```python
# ----- QUERY 5: ORDER BY (Sort results) -----
# Show all enrollments sorted by grade (highest first)
# DESC = descending (highest to lowest)
# ASC = ascending (lowest to highest)
print("\n--- ENROLLMENTS SORTED BY GRADE (Highest First) ---")
run_query("SELECT student_id, course_id, final_grade FROM enrollments ORDER BY final_grade DESC")
```

```python
# ----- QUERY 6: LIMIT (Show only top N results) -----
# Top 3 highest grades
print("\n--- TOP 3 HIGHEST GRADES ---")
run_query("SELECT student_id, course_id, final_grade FROM enrollments ORDER BY final_grade DESC LIMIT 3")
```

```python
# ----- QUERY 7: COUNT, AVG, MAX, MIN (Aggregate functions) -----
# How many students? What's the average grade?
print("\n--- SCHOOL STATISTICS ---")
run_query('''
    SELECT 
        COUNT(*) as total_enrollments,
        ROUND(AVG(final_grade), 1) as average_grade,
        MAX(final_grade) as highest_grade,
        MIN(final_grade) as lowest_grade
    FROM enrollments
''')
# ROUND(..., 1) = round to 1 decimal place
# as = rename the column in the output
```

```python
# ----- QUERY 8: GROUP BY (Summarize by category) -----
# Average grade per course
print("\n--- AVERAGE GRADE PER COURSE ---")
run_query('''
    SELECT course_id, 
           COUNT(*) as num_students, 
           ROUND(AVG(final_grade), 1) as avg_grade
    FROM enrollments
    GROUP BY course_id
''')
```

```python
# ----- QUERY 9: JOIN (Combine two tables — MOST IMPORTANT!) -----
# Show student names WITH their grades
# JOIN connects tables using a shared column (student_id)
print("\n--- STUDENT NAMES WITH THEIR GRADES ---")
run_query('''
    SELECT 
        s.first_name, 
        s.last_name, 
        c.course_name, 
        e.final_grade
    FROM enrollments e
    JOIN students s ON e.student_id = s.student_id
    JOIN courses c ON e.course_id = c.course_id
    ORDER BY e.final_grade DESC
''')
# e, s, c are "aliases" (nicknames) so we don't type full table names
# ON = the matching column between the two tables
```

```python
# ----- QUERY 10: JOIN + WHERE (Filter joined data) -----
# Find students who are FAILING (grade < 50)
print("\n--- FAILING STUDENTS (GRADE < 50) ---")
run_query('''
    SELECT 
        s.first_name, 
        s.last_name, 
        c.course_name, 
        e.final_grade
    FROM enrollments e
    JOIN students s ON e.student_id = s.student_id
    JOIN courses c ON e.course_id = c.course_id
    WHERE e.final_grade < 50
''')
```
