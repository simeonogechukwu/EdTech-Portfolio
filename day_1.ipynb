```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
df
```

```python
print(f"Rows: {len(df)}")
```

```python
print("\nFirst 5 rows:")
print(df.head())
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['total_bill'],
    bins=20,
    color='blue',
    edgecolor='black'
)

plt.title('Bill Distribution')
plt.xlabel('Total Bills ($)')
plt.ylabel('Count')

plt.show()
```

```python
df.groupby('day')['total_bill'].mean().plot(
    kind='bar',
    color='green'
)

plt.title('Average Bill by Day')
plt.ylabel('Average ($)')

plt.show()
```

```python
plt.scatter(
    df['total_bill'],
    df['tip'],
    alpha=0.5
)
plt.title('Bill vs Tip')
plt.xlabel('Total bill ($)')
plt.ylabel('Tip ($)')

plt.show()
```

```python
# Question One: What is the average tip?
avg_tip = df['tip'].mean()
print(f"Average tip: ${avg_tip:.2f}")
#f"text {variable}" = formatted strings (inserts variable value)
```

```python
#Question Two: Do people tip more during weekends?
weekend_tips = df[df['day'].isin(['Sat', 'Sun'])] ['tip'].mean() #df[df['day'].isin(['Sat', 'Sun'])]['tip'].mean()

weekday_tips = df[~df['day'].isin(['Sat', 'Sun'])] ['tip'].mean()

print(f"Weekend tips: ${weekend_tips:.2f}")
print(f"Weekday tips: ${weekday_tips:.2f}")

```

```python
#Question three: Does bill size affect the tip percentage?
#We have to create a new column called tip percent
#Formula = (tip / total_bill) * 100

df['tip_percent'] = (df['tip'] / df['total_bill']) * 100 #df['tip_percent'] = (df['tip'] / df['total_bill']) * 100
print(f"Average tip percentage: {df['tip_percent'].mean():.1f}%")
```

```python
print("\nData with tip percentage:")
print(df[['total_bill', 'tip', 'tip_percent']].head())
```
